# load merges

```python
def convert_merge_result(res):
    """
    Convert a merge string result into a tuple of bytes.
    """
    res = res.rstrip('\n')
    res = res.split(' ')
    if len(res) > 2:
        if res[0]:
            res = [res[0], ' ']
        elif res[-1]:
            res = [' ', res[-1]]
        else:
            pass

    return tuple(bytes(c.encode('utf-8')) for c in res)
```

```python
merges_bytes = list(map(convert_merge_result, merges))
```
