```python
import json
```

```python
def _build_vocab(base_vocab_size, merges=None, special_tokens=None):
    base_vocab_size = base_vocab_size
    vocab = {k: bytes([k]) for k in range(base_vocab_size)}
    # base_vocab_size = vocab_size - 1
    vocab_index = base_vocab_size - 1

    if merges:
        for m in merges:
            vocab_index += 1
            vocab[vocab_index] = b"".join(c for c in m)  # m.encode('utf-8')

    if  special_tokens:
        for token in special_tokens:
            vocab_index += 1
            vocab[vocab_index] = token.encode("utf-8")

    return vocab
```

```python
vocab = _build_vocab(base_vocab_size=256)
```

```python
vocab
```

I cannot save this dict to a json since it contains bytes

# Create a mapping for storing vocab

```python
def convert(v: bytes):
    # Convert a byte sequence to a UTF-8 string
    v_int = list(v)
    v_list = list(map(chr, v_int))
    v_dec = ''.join(v_list)
    return v_dec
```

```python
vocab_to_store = {k: convert(v) for k, v in vocab.items()}
```

```python
vocab_to_store
```

invert the vocab

```python
inv_vocab_to_store = {v: k for k, v in vocab_to_store.items()}
```

```python
with open('test/inv_vocab_to_store.json', 'w') as f:
    json.dump(inv_vocab_to_store, f)
```

# Now I load the saved vocab

```python
with open('test/inv_vocab_to_store.json', 'r') as f:
    inv_vocab_stored = json.load(f)
```

# revert the vocab

# now i want to come back to the original vocab

```python
def map_to_bytes(vocab_char):
    temp = list(map(ord, vocab_char))
    return bytes(temp)
```

```python
vocab_recreated = {k: map_to_bytes(v) for k, v in vocab_stored.items()}
```

```python
assert vocab_recreated == vocab, "Recreated vocabulary does not match the original vocabulary"
```
