```python
with open("resources/raw_text.txt", "r") as f:
    texts = [line.strip() for line in f.readlines()]
texts
```

```python
texts = ["low low low low low", "lower lower widest widest widest", "newest newest newest newest newest newest"]
```

```python
def _pre_tokenize(texts):
    tokens_list = [c.split() for c in texts]
    tokens_list = [t for c in tokens_list for t in c]
    return tokens_list
```

```python
pre_tokens = _pre_tokenize(texts)
```

```python
pre_tokens
```

```python
# count pre tokens
def count_pre_tokens(pre_tokens: list[str]):
    count_dict = {}
    for t in pre_tokens:
        if t in count_dict.keys():
            count_dict[t] += 1
        else:
            count_dict[t] = 1

    return count_dict
```

```python
pre_tokens_freq = count_pre_tokens(pre_tokens)
pre_tokens_freq
```

```python
def get_pair_freq(freq_dict: dict):
    pairs_freq = {}
    for key, value in freq_dict.items():
        for first, second in zip(key, key[1:]):
            pair = (first, second)
            if pair in pairs_freq:
                pairs_freq[pair] += value
            else:
                pairs_freq[pair] = value
    return pairs_freq
```

```python
def merge_pair(freq_dict: dict, top_pair):
    new_freq_dict = {}
    for key, value in freq_dict.items():
        if top_pair in zip(key, key[1:]):
            i = 0
            new_key = []
            # print("found")
            while i < len(key):
                if key[i] == top_pair[0] and key[i + 1] == top_pair[1]:
                    new_key.append(top_pair[0] + top_pair[1])
                    i += 2
                else:
                    new_key.append(key[i])
                    i += 1
            new_freq_dict[tuple(new_key)] = value
        else:
            new_freq_dict[key] = value
    return new_freq_dict
```

```python
freq_dict = merge_pair(freq_dict, top_pair)
freq_dict
```
