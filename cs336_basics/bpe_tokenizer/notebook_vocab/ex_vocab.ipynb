```python
def _build_vocab(base_vocab_size, merges=None, special_tokens=None):
    base_vocab_size = base_vocab_size
    vocab = {k: bytes([k]) for k in range(base_vocab_size)}
    # base_vocab_size = vocab_size - 1
    vocab_index = base_vocab_size - 1

    if merges:
        for m in merges:
            vocab_index += 1
            vocab[vocab_index] = b"".join(c for c in m)  # m.encode('utf-8')

    if special_tokens:
        for token in special_tokens:
            vocab_index += 1
            vocab[vocab_index] = token.encode("utf-8")

    return vocab
```

```python
vocab = _build_vocab(base_vocab_size=256)
```

# create a mapping for storing vocab

```python
bs = list(range(ord("!"), ord("~") + 1)) + list(range(ord("¡"), ord("¬") + 1)) + list(range(ord("®"), ord("ÿ") + 1))
cs = bs[:]
# now get the representations of the other 68 integers that do need shifting
# each will get mapped chr(256 + n), where n will grow from 0...67 in the loop
# Get printable representations of the remaining integers 68 integers.
n = 0
for b in range(2**8):
    if b not in bs:
        # If this integer isn't in our list of visually-representable
        # charcters, then map it to the next nice character (offset by 256)
        bs.append(b)
        cs.append(2**8 + n)
        n += 1
characters = [chr(n) for n in cs]
d = dict(zip(bs, characters))
```
