# Decoding bytes to string

```python
from urllib.request import urlopen

url = "https://github.com/"

with urlopen(url) as response:
    raw_bytes: bytes = response.read()

print("Bytes:", raw_bytes[:100])
print("String:", raw_bytes[:100].decode())
```

The .decode() method takes two optional arguments: <br>

Argument	Default	Meaning <br>
- encoding	"utf-8"	The encoding with which to decode the bytes
- errors	"strict"	The handling scheme to use for decoding errors

```python
b"d\xe9j\xe0 vu".decode("utf-8", errors="ignore")
'dj vu'
```
