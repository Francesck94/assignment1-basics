```python
import torch
```

```python
from einops import einsum, rearrange, reduce, repeat
```

```python
def xavier_init(in_features, out_features):
    std = torch.sqrt(torch.tensor(2)/(in_features + out_features))
    return torch.clip(torch.normal(mean=0, std=std, size=(out_features, in_features)), -3*std, 3*std)
```

```python
in_feature = 2
out_feature = 3
weights = xavier_init(in_feature, out_feature)
```

```python
batch_size = 4
x = torch.randn((in_feature, batch_size))
```
